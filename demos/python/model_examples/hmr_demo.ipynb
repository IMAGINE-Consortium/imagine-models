# HMR Field

**Reference:** [Harari, Mollerach & Roulet (1999)](https://arxiv.org/abs/astro-ph/9906309)  
**Based on:** [hammurabi v3.01](https://sourceforge.net/projects/hammurabicode/)  
Deviations from the publication: see the [README](../../../README.md#deviations-from-the-publications).

```python
import ImagineModels as img
from ImagineModels import plot_slice
```

#### 1. Visualizing the model

```python
model = img.HMRMagneticField()
shape = [200, 200, 40]  # necessarily a list of 3 ints!
increment = [0.20, 0.20, 0.20]
refpoint = [-20.0, -20.0, -4.0]
vmin = -6
vmax = 6
```

```python
model_on_grid = model.evaluate(img.RegularGrid(shape=shape, reference_point=refpoint, increment=increment))
plot_slice(model_on_grid, 2, shape, refpoint, increment, vmin, vmax, vec_dim=0, field_name="HMR 1999")
```

```python
plot_slice(model_on_grid, 2, shape, refpoint, increment, vmin, vmax, vec_dim=1, field_name="HMR 1999")
```

```python
plot_slice(
    model_on_grid, 2, shape, refpoint, increment, vmin, vmax, vec_dim=2, field_name="HMR 1999"
)  # no field in z-direction
```

```python
plot_slice(model_on_grid, 2, shape, refpoint, increment, vmin, vmax, quiver=True, field_name="HMR 1999", amplitude=True)
```

```python
plot_slice(model_on_grid, 1, shape, refpoint, increment, vmin, vmax, vec_dim=0, field_name="HMR 1999")
```

```python
plot_slice(model_on_grid, 1, shape, refpoint, increment, vmin, vmax, vec_dim=1, field_name="HMR 1999")
```

```python
plot_slice(model_on_grid, 1, shape, refpoint, increment, vmin, vmax, vec_dim=2, field_name="HMR 1999")
```

```python
plot_slice(
    model_on_grid, 1, shape, refpoint, increment, vmin, vmax, field_name="HMR 1999", amplitude=True
)  # BSS-S model
```
