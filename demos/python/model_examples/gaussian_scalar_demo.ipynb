```python
import numpy as np

import ImagineModels as img
from ImagineModels import plot_slice
```

#### 1. Visualizing the model

```python
model = img.GaussianScalarField()
shape = [100, 100, 20]  # necessarily a list of 3 ints!
increment = [0.20, 0.20, 0.20]
refpoint = [-10.0, -10.0, -2.0]
vmin = -1.0
vmax = 1.0

model.sigma = 1.0
model.spectral_offset = 0.01
model.spectral_slope = 2
model.apply_spectrum = True

seed = 23

model_on_grid = model.sample(img.RegularGrid(shape=shape, reference_point=refpoint, increment=increment), seed=seed)
print(np.std(model_on_grid))
plot_slice(model_on_grid, 0, shape, refpoint, increment, vmin, vmax, label="")
```

```python
plot_slice(model_on_grid, 1, shape, refpoint, increment, vmin, vmax, label="")
```

```python
plot_slice(model_on_grid, 2, shape, refpoint, increment, vmin, vmax, label="")
```
