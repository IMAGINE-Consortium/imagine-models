# TF17 Field

**Reference:** [Terral & Ferrière (2017)](https://arxiv.org/abs/1611.10222)  
**Based on:** [CRPropa](https://github.com/CRPropa/CRPropa3)  
Deviations from the publication: see the [README](../../../README.md#deviations-from-the-publications).

Only the halo was fitted to data, which leads to very strong field strengths and unexpected features in the disk fields; the halo fields can diverge at large r/z. The model is better viewed as a mathematical exercise than as a model for, e.g., cosmic-ray propagation.

```python
import ImagineModels as img
from ImagineModels import plot_slice
```

#### 1. Visualizing the model

```python
model = img.TFMagneticField()
shape = [200, 200, 40]  # necessarily a list of 3 ints!
increment = [0.20, 0.20, 0.20]
refpoint = [-20.0, -20.0, -4.0]
vmin = -100
vmax = 100.0

# The model has different disk and halo fields, that can be changed in the following way:
model.set_model("Bd1", "C1")  # disk: Ad1, Bd1 or Dd1; halo: C0 or C1
```

```python
model_on_grid = model.evaluate(img.RegularGrid(shape=shape, reference_point=refpoint, increment=increment))
plot_slice(model_on_grid, 2, shape, refpoint, increment, vmin, vmax, vec_dim=0, field_name="TF17")
```

```python
plot_slice(model_on_grid, 2, shape, refpoint, increment, vmin, vmax, vec_dim=1, field_name="TF17")
```

```python
plot_slice(model_on_grid, 2, shape, refpoint, increment, vmin, vmax, vec_dim=2, field_name="TF17")
# model has no z component
```

```python
plot_slice(model_on_grid, 2, shape, refpoint, increment, vmin, vmax, field_name="TF17", quiver=True, amplitude=True)
```

```python
plot_slice(model_on_grid, 1, shape, refpoint, increment, vmin, vmax, vec_dim=0, field_name="TF17")
```

```python
plot_slice(model_on_grid, 1, shape, refpoint, increment, vmin, vmax, vec_dim=1, field_name="TF17")
```

```python
plot_slice(model_on_grid, 1, shape, refpoint, increment, vmin, vmax, vec_dim=2, field_name="TF17")
```

```python
plot_slice(model_on_grid, 1, shape, refpoint, increment, vmin, vmax, field_name="TF17", quiver=True, amplitude=True)
```
