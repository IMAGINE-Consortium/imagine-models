# Jaffe Field

**Reference:** [Jaffe et al. (2010)](https://ui.adsabs.harvard.edu/abs/2010MNRAS.401.1013J/abstract)  
**Based on:** [hammurabiX](https://github.com/hammurabi-dev/hammurabiX)  
Deviations from the publication: see the [README](../../../README.md#deviations-from-the-publications).

```python
import ImagineModels as img
from ImagineModels import plot_slice
```

#### 1. Visualizing the model

```python
model = img.JaffeMagneticField()
shape = [200, 200, 40]  # necessarily a list of 3 ints!
increment = [0.20, 0.20, 0.20]
refpoint = [-20.0, -20.0, -4.0]
vmax = 5.0
vmin = -5.0

# The model can be used either with a ring or bar feature (default) in the innermost region. These can be changed with:
model.ring = True
model.bar = False
```

```python
model_on_grid = model.evaluate(img.RegularGrid(shape=shape, reference_point=refpoint, increment=increment))
plot_slice(model_on_grid, 2, shape, refpoint, increment, vmin, vmax, vec_dim=0, field_name="Jaffe et al 2010")
```

```python
plot_slice(model_on_grid, 2, shape, refpoint, increment, vmin, vmax, vec_dim=1, field_name="Jaffe et al 2010")
```

```python
plot_slice(model_on_grid, 2, shape, refpoint, increment, vmin, vmax, vec_dim=2, field_name="Jaffe et al 2010")
# The model has no z-component
```

```python
plot_slice(
    model_on_grid, 2, shape, refpoint, increment, vmin, vmax, field_name="Jaffe et al 2010", quiver=True, amplitude=True
)
```

```python
plot_slice(model_on_grid, 1, shape, refpoint, increment, vmin, vmax, vec_dim=0, field_name="Jaffe et al 2010")
```

```python
plot_slice(model_on_grid, 1, shape, refpoint, increment, vmin, vmax, vec_dim=1, field_name="Jaffe et al 2010")
```

```python
plot_slice(model_on_grid, 1, shape, refpoint, increment, vmin, vmax, vec_dim=2, field_name="Jaffe et al 2010")
```

```python
plot_slice(model_on_grid, 1, shape, refpoint, increment, vmin, vmax, field_name="Jaffe et al 2010", amplitude=True)
```
