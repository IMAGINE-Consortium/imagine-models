# Han et al Field

**-Reference:** [Han et al. (2018)](https://iopscience.iop.org/article/10.3847/1538-4365/aa9c45)

**-Original Implementation:** /

**-Default parameters constrained using (see reference publication):** measurements of Faraday rotation for 477 pulsars + previous measurements of pulsars and extragalactic sources

```python
import ImagineModels as img
from ImagineModels import plot_slice
```

#### 1. Visualizing the model

```python
model = img.HanMagneticField()
shape = [200, 200, 40]  # necessarily a list of 3 ints!
increment = [0.20, 0.20, 0.20]
refpoint = [-20.0, -20.0, -4.0]
vmin = -5.0  # field strength is weaker than most others
vmax = 5.0
```

```python
model_on_grid = model.evaluate(img.RegularGrid(shape=shape, reference_point=refpoint, increment=increment))
plot_slice(model_on_grid, 2, shape, refpoint, increment, vmin, vmax, vec_dim=0, field_name="Han et al 2018")
```

```python
plot_slice(model_on_grid, 2, shape, refpoint, increment, vmin, vmax, vec_dim=1, field_name="Han et al 2018")
```

```python
plot_slice(model_on_grid, 2, shape, refpoint, increment, vmin, vmax, vec_dim=2, field_name="Han et al 2018")
# no field in z direction
```

```python
plot_slice(
    model_on_grid, 2, shape, refpoint, increment, vmin, vmax, field_name="Han et al 2018", quiver=True, amplitude=True
)
```

```python
plot_slice(model_on_grid, 1, shape, refpoint, increment, vmin, vmax, vec_dim=0, field_name="Han et al 2018")
```

```python
plot_slice(model_on_grid, 1, shape, refpoint, increment, vmin, vmax, vec_dim=1, field_name="Han et al 2018")
```

```python
plot_slice(model_on_grid, 1, shape, refpoint, increment, vmin, vmax, vec_dim=2, field_name="Han et al 2018")
```

```python
plot_slice(model_on_grid, 1, shape, refpoint, increment, vmin, vmax, field_name="Han et al 2018", amplitude=True)
```
